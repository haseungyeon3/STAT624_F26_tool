```python
from datetime import datetime, timezone

 
# creates "students" automatically when data is inserted.
students = db["students"]

# Optional: explicitly create a collection only if it does not exist.
if "students" not in db.list_collection_names():
    db.create_collection("students")
    print("Created collection: students")

# 3. Prepare many MongoDB documents
student_documents = [
    {
        "studentId": 1001,
        "name": "Alice Johnson",
        "email": "alice@example.com",
        "program": "Computer Science",
        "active": True,
        "createdAt": datetime.now(timezone.utc)
    },
    {
        "studentId": 1002,
        "name": "Bob Smith",
        "email": "bob@example.com",
        "program": "Data Analytics",
        "active": True,
        "createdAt": datetime.now(timezone.utc)
    },
    {
        "studentId": 1003,
        "name": "Carol Lee",
        "email": "carol@example.com",
        "program": "Software Engineering",
        "active": False,
        "createdAt": datetime.now(timezone.utc)
    }
]

# 4. Insert all documents at once
result = students.insert_many(student_documents)

print("Inserted IDs:", result.inserted_ids)
print("Total documents:", students.count_documents({}))

# 5. Verify
for document in students.find():
    print(document)
```
